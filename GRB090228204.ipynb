## Examining the GBM catalog for GRB090228204

```python
gbm_catalog = FermiGBMBurstCatalog()
gbm_catalog.query_sources("GRB090228204")
```

```python
# define source location and trigger for the GRB
ra = 106.800
dec = -24.300
trigger = "GRB090228204"
```

```python
grb_info = gbm_catalog.get_detector_information()[trigger]

gbm_detectors = grb_info["detectors"] #triggered detectors info
background_interval = grb_info["background"]["full"] #off times
best_fit_model = grb_info["best fit model"]["fluence"] #best fit model for the GRB in the catalog
model_catalog = gbm_catalog.get_model(best_fit_model, "fluence")[trigger]
```

## Downloading the data

```python
# only for the triggered GBM detectors
dload = download_GBM_trigger_data("bn090228204", detectors=gbm_detectors)
```

## Temporal binning as defined in Burgess et. al. 2019 (https://doi.org/10.1093/mnras/stz2589)

```python
# on times
source = ['-0.009--0.002', '-0.002-0.019','0.019-0.034','0.034-0.081','0.081-0.101'
          , '0.101-0.155','0.155-0.260','0.260-0.433']
```

## Measurement for time bin 1

```python
fluence_plugins = []
time_series = {}
for det in gbm_detectors:

    ts_cspec = TimeSeriesBuilder.from_gbm_cspec_or_ctime(
        det, cspec_or_ctime_file=dload[det]["cspec"], rsp_file= dload[det]["rsp"]
    )

    ts_cspec.set_background_interval(*background_interval.split(',')) #performs a polynomial fit on the selected interval

    # creating a time series object (that includes the lightcurve) for only the selected detectors
    ts_tte = TimeSeriesBuilder.from_gbm_tte(
        det,
        tte_file= dload[det]["tte"],
        rsp_file= dload[det]["rsp"]
    )
    
    ts_tte.set_background_interval(*background_interval.split(','))

    time_series[det] = ts_tte

    ts_tte.set_active_time_interval(source[0])

    ts_tte.view_lightcurve(-10, 10)

    fluence_plugin = ts_tte.to_spectrumlike()

    if det.startswith("b"):

        fluence_plugin.set_active_measurements("250-30000") #selected energy range for the BGO detectors

    else:

        fluence_plugin.set_active_measurements("9-900") #selected energy range for the NaI detectors

    #fluence_plugin.rebin_on_background(1.0)

    fluence_plugins.append(fluence_plugin)
```

```python
#look at spectrum
for i in range(len(gbm_detectors)):
    fluence_plugins[i].view_count_spectrum();
```

## Defining the model

```python
# using a Cutoff Powerlaw model to fit the spectrum
cpl = Cutoff_powerlaw(piv=100)

# define spatial model (point source)
ptsrc = PointSource(trigger, ra, dec, spectral_shape=cpl)

# combine to a spatial-spectral model
model = Model(ptsrc)
```

```python
# check out complete model with free and fixed parameters
model.display(complete=True)
```

## Setting priors for the model

```python
# for a Bayesian fit, we need to define priors
# that is, knowledge (or ignorance) about the parameters
# this needs to be done for every free parameter
# if we don't know anything about the parameters, we define broad ranges

#amplitude
model.GRB090228204.spectrum.main.Cutoff_powerlaw.K.bounds = (1e-10,1e10)
model.GRB090228204.spectrum.main.Cutoff_powerlaw.K.prior = Log_normal(mu = -2, sigma=5)

#index
model.GRB090228204.spectrum.main.Cutoff_powerlaw.index.prior = Gaussian(mu=-1, sigma=0.5)

#cutoff energy
model.GRB090228204.spectrum.main.Cutoff_powerlaw.xc.prior = Truncated_gaussian(lower_bound=10, upper_bound=10000,mu=200, sigma=300)
```

```python
from threeML.bayesian.sampler_base import SamplerBase
```

## Cloning the model and setting up the bayesian analysis class

```python
#clone the original model to not affect it from the fit
new_model = clone_model(model)

#performs bayesian analysis by passing model and data
bayes = BayesianAnalysis(new_model, DataList(*fluence_plugins))

# share spectrum gives a linear speed up when
# spectrumlike plugins have the same RSP input energies
bayes.set_sampler("emcee", share_spectrum=True)
```

## Running the sampler

```python
bayes.sampler.setup(n_iterations=200)
bayes.sample()
```

```python
#sets model parameters to the mean of the distribution
bayes.restore_median_fit()

#display the fit results
bayes.results.display()
```

## Examining the fitted model

```python
fig = display_spectrum_model_counts(bayes, min_rate=10)
fig.axes[0].set_ylim(1e-9)
```

```python
fluxes = bayes.results.get_flux(10 * u.keV , 1*u.MeV)
fluxes
```

```python
FE = 32.14e-6 * u.erg/u.cm**2/u.s
Epeak = 339.6 * u.keV
alpha = -0.77
Ecut = Epeak / (2 + alpha)
Epiv = 100 *u.keV
Emin = 10*u.keV
Emax = 1000*u.keV


K = FE / (Ecut**(2+alpha) * Epiv**(-alpha) * ( gammaincc(2+alpha,(Emax/Ecut).value) - gammaincc(2+alpha,(Emin/Ecut).value) ))
K = K.to(1/u.keV/u.s/u.cm**2)




```

```python
K, Ecut
```

## Spectral parameters from Burgess et. al. 2019

```python
cpl = Cutoff_powerlaw(piv=100,xc = Ecut.value, index = alpha, K = -(K.value))
ptsrc = PointSource(trigger, ra, dec, spectral_shape=cpl)
model_burgess = Model(ptsrc)
#model_burgess.display(complete=True)
```

```python
conversion = u.Unit("keV2/(cm2 s keV)").to("erg2/(cm2 s keV)")
energy_grid = np.logspace(1, 4, 1000) * u.keV
vFv_burgess = (energy_grid**2 * model_burgess.get_point_source_fluxes(0, energy_grid)).to(
    "erg2/(cm2 s keV)"
)
# vFv_catalog = (energy_grid**2 * model_catalog.get_point_source_fluxes(0, energy_grid)).to(
#     "erg2/(cm2 s keV)"
# )
```
